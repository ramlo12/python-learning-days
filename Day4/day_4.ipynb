##Filling Numerical Data

```python
# Hel celceliska Quantity-ka, adigoo iska indhatiraya missing value
Quantity_mean = data["Quantity"].mean()

Quantity_mean
```

```python
# Hel median-ka Quantity, iyadoo missing values-ka la iska indhatirayo
Quantity_median = data["Quantity"].median()

Quantity_median
```

```python
# Buuxi Quantity-da maqan adigoo isticmaalaya mean-ka Quantity
data["Quantity"] = data["Quantity"].fillna(Quantity_mean)
print(data["Quantity"])
```

```python
# Hubi inta missing value ee hadda ku jirta Quantity
data["Quantity"].isna().sum()
```

```python
# Hel mean-ka Unit_Price, iyadoo missing value-ka la iska indhatirayo
Unit_Price_mean = data["Unit_Price"].mean()

Unit_Price_mean
```

```python
# Hel median-ka Unit_Price
Unit_Price_median = data["Unit_Price"].median()

Unit_Price_median
```

```python
# Buuxi Unit_Price-da maqan adigoo isticmaalaya mean-ka
data["Unit_Price"] = data["Unit_Price"].fillna(Unit_Price_mean)
```

```python
# Hubi inta missing value ee hadda ku jirta Unit_Price
data["Unit_Price"].isna().sum()
```

## fill Categorical Data

```python
# Hel mode-ka Product
Product_mode = data["Product"].mode()[0]

Product_mode
```

```python
# Buuxi Product-ka maqan adigoo isticmaalaya mode-ka Product
data["Product"] = data["Product"].fillna(Product_mode)
```

```python
data["Product"].isna().sum()
```

```python
# Hel magaalada ugu badan (mode) ee City
City_mode = data["City"].mode()[0]

City_mode
```

```python
# Buuxi City-ga maqan adigoo isticmaalaya mode-ka City
data["City"] = data["City"].fillna(City_mode)
```

```python
# Hubi inta missing value ee hadda ku jirta City
data["City"].isna().sum()
```

```python
# Hel Customer_Type-ka ugu badan (mode)
Customer_Type_mode = data["Customer_Type"].mode()[0]

Customer_Type_mode
```

```python
# Buuxi Customer_Type-ka maqan adigoo isticmaalaya mode-ka
data["Customer_Type"] = data["Customer_Type"].fillna(Customer_Type_mode)
```

```python
# Hubi inta missing value ee hadda ku jirta Customer_Type
data["Customer_Type"].isna().sum()
```
