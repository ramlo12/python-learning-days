# Filtering means selecting only observations
# that satisfy a specific condition
##Interpretation:
Filtering waxaa loola jeedaa inaan xogta ka soo saarno kaliya rows-ka buuxinaya shuruud gaar ah.





 Numerical Conditions

```python
# Select customer whose Premium is less than 110
# < means less than
data[data['Premium'] < 110]
```

##Categorical Conditions

```python
# Select only customar whose Policy_Type is Travel
# == means equal to
data[data['Policy_Type'] == "Travel"]
```

## Filtering Multiple Categories
Interpretation:
Halkan waxaan rabnaa inaan soo saarno policies-ka Claim_Status-koodu yahay Open ama Closed.

data['Claim_Status'] → waxaan dooranaynaa column-ka Claim_Status
.isin() → wuxuu hubinayaa haddii value-ku ku jiro list-ka

```python
# Select rows where Claim_Status is either Open or Closed
# isin() checks whether a value belongs to a list of values
data[data['Claim_Status'].isin(['Approved', 'Rejected'])]
```

```python
# Select customers younger than 30
# AND whose claim status is Open
# & means AND
data[(data['Premium'] < 400) & (data['Policy_Type'] == 'Property')]
```

##Sorting Data

Interpretation:
Halkan waxaan xogta ku kala saaraynaynaa da'da customer-ka.

data.sort_values() → waxaa loo isticmaalaa sorting.
'Customer_Age' → waa column-ka aan ku kala saaraynayno.
Default ahaan → wuxuu isticmaalaa ascending order.

```python
# Sort the data according to the Customer_Age column
# By default, Pandas sorts in ascending order
data.sort_values('Customer_Age')
```

```python
# Sort Customer_Age from the smallest value to the largest value
# ascending=True means ascending order
data.sort_values('Customer_Age', ascending=True)
```

```python
# Sort Customer_Age from the largest value to the smallest value
# ascending=False means descending order
data.sort_values('Customer_Age', ascending=False)
```

```python
# Sort the data first by Policy_Type
# Then sort by Claim_Status
data.sort_values(by=['Policy_Type', 'Claim_Status'])
```
