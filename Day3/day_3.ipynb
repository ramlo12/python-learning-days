```python
import pandas as pd
```

```python
data = pd.read_excel("/content/drive/MyDrive/Colab Notebooks/Supermarket_Practice_Data.xlsx")
```

```python
data = data.rename(columns={
    "Invoice_ID": "InvoiceID"
})
```

##Using replace

```python
data['Payment_Method'].value_counts()
```

```python
data["Payment_Method"]=data["Payment_Method"].replace("evc plus","EVC Plus")
```

```python
data.columns
```

```python
data['City'].value_counts()
```

```python
data["City"]=data["City"].replace("Mogadisho","Mogadishu")
```

```python
data['Category'].value_counts()
```

```python
# Replace the value "grocery" with "Grocery" in the Category column
data["Category"] = data["Category"].replace("grocery", "Grocery")
```

##Numerical Replacement

```python
data["Quantity"] = data["Quantity"].replace(
{ 1: "Low",
 2: "Medium",
3: "more" })
```

```python
data.head()
```

##Removing Unnecessary Columns

```python
data.drop(
columns=["InvoiceID"]
)
```

##hanging Data Types

```python
data.dtypes
```

```python
data["Quantity"] = pd.to_numeric(data["Quantity"],
errors='coerce').astype('Int64')
```

##str.strip()
 The .str.strip() method removes unnecessary spaces from the beginning and end of text values

```python
data["Category"] = data["Category"].str.strip()
```

##str.lower() & .str.upper()
The .str.lower()method converts text to lowercase

```python
data["Category"] = data["Category"].str.title()
```
